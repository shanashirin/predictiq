```python
outlier_columns=[    'Air temperature [K]', 'Process temperature [K]',
    'Rotational speed [rpm]', 'Torque [Nm]', 'Tool wear [min]'
]
for col in outlier_columns:
    q1=df[col].quantile(0.25)
    q3=df[col].quantile(0.75)
    iqr=q3-q1
    lower=q1-1.5*iqr
    upper=q3+1.5*iqr
    count=((df[col]<lower)|(df[col]>upper)).sum()
    print(col, 'possible outliers =', count, '| lower =', round(lower, 2), '| upper =', round(upper, 2))
```

```python
outlier_limits = {
    "Rotational speed [rpm]": (1139.5, 1895.5),
    "Torque [Nm]": (12.8, 67.2)
}

for col, (lower, upper) in outlier_limits.items():

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"{col}: Remaining outliers = {len(outliers)}")
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.get_dummies(df,columns=["Type"],drop_first=True,dtype=int)
```

```python
remove_columns = [
    'Machine failure', 'UDI', 'Product ID',
    'TWF', 'HDF', 'PWF', 'OSF', 'RNF'
]
remove_columns = [col for col in remove_columns if col in df.columns]
x=df.drop(remove_columns,axis="columns")
y=df["Machine failure"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
xtrain.shape
```

```python
ytrain.shape
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
mn=MinMaxScaler()
```

```python
xtrain_scaled=mn.fit_transform(xtrain)
```

```python
xtest_scaled=mn.transform(xtest)
```

```python
ytrain.value_counts()
```

```python
from imblearn.over_sampling import SMOTE
```

```python
s=SMOTE(random_state=42)
```

```python
xtrain_balanced,ytrain_balanced=s.fit_resample(xtrain_scaled,ytrain)
```

```python
ytrain_balanced.value_counts()
```

```python
sns.countplot(x=ytrain_balanced)
plt.title('Training Target After SMOTE')
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neural_network import MLPClassifier
```

```python
def get_score(algorithm,xtrain,xtest,ytrain,ytest):
    algorithm.fit(xtrain,ytrain)
    return algorithm.score(xtest,ytest)
```

```python
lo=LogisticRegression(max_iter=1000,class_weight="balanced",random_state=42)
get_score(lo,xtrain_balanced,xtest_scaled,ytrain_balanced,ytest)

```

```python
dt = DecisionTreeClassifier(class_weight='balanced', random_state=42)
get_score(dt,xtrain_balanced,xtest_scaled,ytrain_balanced,ytest)
```

```python
rf = RandomForestClassifier(n_estimators=150, class_weight='balanced', random_state=42, n_jobs=-1)
get_score(rf, xtrain_balanced,xtest_scaled,ytrain_balanced,ytest)
```

```python
ann = MLPClassifier(hidden_layer_sizes=(32, 16), max_iter=300, early_stopping=True, random_state=42)
get_score(ann,xtrain_balanced,xtest_scaled,ytrain_balanced,ytest)
```

```python

import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

models = {
    "Logistic Regression": lo,
    "Decision Tree": dt,
    "Random Forest": rf,
    "ANN": ann
}

results = []

for name, model in models.items():
    y_pred = model.predict(xtest_scaled)
    y_prob = model.predict_proba(xtest_scaled)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(ytest, y_pred),
        "Precision": precision_score(
            ytest, y_pred, zero_division=0
        ),
        "Recall": recall_score(
            ytest, y_pred, zero_division=0
        ),
        "F1-score": f1_score(
            ytest, y_pred, zero_division=0
        ),
        "ROC-AUC": roc_auc_score(ytest, y_prob),
        "PR-AUC": average_precision_score(ytest, y_prob)
    })


# Sort models by actual PR-AUC, highest first
results_df = results_df.sort_values(
    by="PR-AUC",
    ascending=False
).reset_index(drop=True)

# Display the actual ranking
print("Model Ranking:")
print(results_df[["Model", "PR-AUC", "Recall", "F1-score"]].round(4))

# Plot the ranked models
import matplotlib.pyplot as plt

results_df.set_index("Model")[
    ["Precision", "Recall", "F1-score", "PR-AUC"]
].plot(kind="bar", figsize=(12, 6))

plt.title("Model Comparison — Ranked by PR-AUC")
plt.ylabel("Score")
plt.xlabel("Model")
plt.ylim(0, 1)
plt.xticks(rotation=15)
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import confusion_matrix,classification_report
```

```python
best_name = results_df.iloc[0]["Model"]

print("Best Model:", best_name)
best_model = models[best_name]

prediction = best_model.predict(xtest_scaled)
probability = best_model.predict_proba(xtest_scaled)[:, 1]

print('Selected model:', best_name)
print(classification_report(ytest, prediction, target_names=['No failure', 'Failure'], zero_division=0))

cm = confusion_matrix(ytest, prediction)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title('Confusion Matrix - ' + best_name)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid = GridSearchCV(
    RandomForestClassifier(class_weight='balanced', random_state=42, n_jobs=-1),
    {
        'n_estimators': [100, 150],
        'max_depth': [None, 10],
        'min_samples_leaf': [1, 3]
    },
    cv=3,
    scoring='average_precision',
    n_jobs=-1
)
```

```python
grid.fit(xtrain_balanced,ytrain_balanced)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
tuned_rf = grid.best_estimator_
tuned_prediction = tuned_rf.predict(xtest_scaled)
tuned_probability = tuned_rf.predict_proba(xtest_scaled)[:, 1]

print('Tuned Random Forest test accuracy:', round(accuracy_score(ytest, tuned_prediction), 4))
print('Tuned Random Forest test precision:', round(precision_score(ytest, tuned_prediction, zero_division=0), 4))
print('Tuned Random Forest test recall:', round(recall_score(ytest, tuned_prediction, zero_division=0), 4))
print('Tuned Random Forest test F1:', round(f1_score(ytest, tuned_prediction, zero_division=0), 4))
print('Tuned Random Forest test ROC-AUC:', round(roc_auc_score(ytest, tuned_probability), 4))
print('Tuned Random Forest test PR-AUC:', round(average_precision_score(ytest, tuned_probability), 4))
```
